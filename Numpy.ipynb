### Display Arrays

```python
numbers = np.array([ 2, 3 , 5, 10])

for val in numbers:
	print (val , end =' ')
```

```python
num2D = np.array( [  [1,3,5,7,9],[2,4,6,8,10] ] )

for row in num2D:
    for val in row :
        print (val , end = ' ')
    print ()
```

## Randomly generated arrays

```python
#rand()  generate random number uniformly distributed
100 * np.random.rand(5, 2)
```

```python
#randn()  generate random number normally distributed
np.random.randn(5, 2)
```

```python
#randint generate random uniformly distributed integer between a low and high value
np.random.randint(low=0, high=100, size=(5, 2))
```
