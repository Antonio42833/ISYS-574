We want to create a list  new_list2 containing all the consecutive even integers between 5 and 14

```python
# Option 1 
new_list2 = []      
for item in range (5,14):
    if item % 2 == 0 :       # We check for even number     
        new_list2.append(item)

new_list2
```

```python
#Option 2 : List Comprehension
new_list2 = [ item for item in range(5,14) if item % 2 == 0]

new_list2
```

### Value Return

```python
def lst_sum(lstA):
    Sum = 0	
    for item in lstA:
        Sum = item + Sum
    return Sum
```

```python
lst_sum([1,2,3,5])  
```

```python
lst_sum([-10,2.5,78,-5.8])  
```

```python
total = lst_sum([1,2,3,5])
print(total)
```

```python
## Class

# we define the class
class NumberPrinter:
    def __init__(self, n):
        self.value = n

    def print_numbers(self):
        for i in range(self.value + 1) :
            print(i)
```

```python

num_printer = NumberPrinter(n = 10 )
```

```python
# we display the values until n
num_printer.print_numbers()
```
