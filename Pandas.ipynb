## Split the DataFrame

```python
#how many rows are in the dataset
row_numbers = cars.shape[0]
print('row numbers =', row_numbers)

#I want 60% ofthem in the train set
n_train = int(row_numbers*0.6)

#I want 20% of them in the validation 
n_val = int(row_numbers*0.20)

print('train until row :', n_train)
print('validate until row:', n_train+n_val)
```

Since it is a very small dataset, we cannot really do the usual 80%/10%/10% split

```python
train_set = cars.iloc[:n_train]
train_set
```

```python
validation_set = cars.iloc[n_train:n_train + n_val]
validation_set
```

```python
test_set = cars.iloc[n_train + n_val:]
test_set
```
